Starting with a given list, count the occurrences of each unique element.  Create a dictionary to store the count of each element.

```python
list1 = [34, 544, 544, 34, 67, 67, 34, 5, 7, 9]
```


````python
list1 = [34, 544, 544, 34, 67, 67, 34, 5, 7, 9]
list1
````

````python


new = {}

for number in list1:  # iterates through the numbers
    if number in new:   # checks the number
        new[number] += 1      # already in dictionary — counts the number
    else:
        new[number] = 1       # first time seen in list add to new list
print("=" * 25)
print(f"{'Number':^10} | {'Count':^8}")
print("=" * 25)

for key, value in new.items():  # iterates and places number in the table and counts
    print(f"{key:^10} | {value:^8}")

print("=" * 25)
````

Given a list, check if each element already exists in a dictionary as a key's value.  If not, delete the element from the list.

```python
list1 = [4, 5, 6, 7, 3, 5, 4, 3]
fruit_counts = {
    "apple" : 4,
    "orange": 8,
    "grapefruit": 3,
    "grapes": 7
}
```


````python
list1 = [4, 5, 6, 7, 3, 5, 4, 3]
fruit_counts = {
    "apple" : 4,
    "orange": 8,
    "grapefruit": 3,
    "grapes": 7
}
print(list1)
print(fruit_counts)
````

````python
list1 = [num for num in list1 if num in fruit_counts.values()]
unique_values = list(set(fruit_counts.values()))
print(list1)
print(unique_values)   # → [3, 4, 7, 8]
````

Given a string input calculate and display the number of digits and letters.

```python
s = "I <3 python"
```

````python
s = "I <3 python"
print(s)
````

````python
letter = sum(1 for char in s if char.isalpha())  #count each char if it is a char.isaplha
digit  = sum(1 for char in s if char.isdigit())  # sums each number if digit

print(f"Letters: {letter}")  #prints letter
print(f"Digits:  {digit}")  # print digit
````
