```
5
```
Output:

```python
@
@@
@@@
@@@@
@@@@@
@@@@
@@@

```


Solution:

````python
# Solution here
def bird(ob):
  ''' this prints a pattern using
  two for loops and a print
'''
  for x in range(1, ob + 1):
    print('@' * x)
  for x in range(ob -1, 0,  -1):
     print('@' * x)

bird(10)
````

# Problem 5



Write a Python function that takes a positive integer (n) as input and iterates through the integers from 0 to n.

Return a list that has
-  the number if it is not a multiple of 3 or 5
-  'Fizz' if the number is a multiple of 3
-  'Buzz' if the number is a multiple of 5
-  'FizzBuzz' if the number is a multiple of 3 and 5

Sample Input:
```
7
```

Sample Return:
```
['FizzBuzz', 1, 2, 'Fizz', 4, 'Buzz', 'Fizz', 7]
```

````python
# Solution here
def fizzbuzz(n):
  ''' creates a for loop appending values to a list called
  results and prints out fizzbuzz if conditions are met
'''
  results = []
  for i in range(1, n + 1):
        if i % 3 == 0 and i % 5 == 0:
            results.append('FizzBuzz')
        elif i % 3 == 0:
            results.append('Fizz')
        elif i % 5 == 0:
            results.append('Buzz')
        else:
            results.append(i)

  return(results)
print(fizzbuzz(21))
````

# Problem 6



Write a Python function to calculate a dog's age in dog's years.  (Input is human age).

Note: For the first two years, a dog year is equal to 10.5 human years. After that, each dog year equals 4 human years.

Input:

```
h_age = 5
```
Returns the string:
```
The dog's age in dog's years is 33
```


````python
# Solution here
def dogz_ages(human_ages):
  ''' This calculates dog years to human years'''
  if human_ages <= 2:
    dog_years = human_ages * 10.5
  else:
    dog_years = (2 * 10.5) + ((human_ages - 2) * 4)
  return dog_years
print(dogz_ages(15))
````
