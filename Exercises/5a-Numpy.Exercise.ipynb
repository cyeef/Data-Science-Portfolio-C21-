## Problem 4: stats


Find the min, max, mean, sample standard deviation, and population standard deviation for the array below.

```python
arr = np.random.normal(5,10,10000)
```


````python
arr = np.random.normal(5,10,10000)
arr = np.random.normal(5, 10, 10000)

print('='*40)
print(f"{'Min':<25} {arr.min():.4f}")
print(f"{'Max':<25} {arr.max():.4f}")
print(f"{'Mean':<25} {arr.mean():.4f}")
print(f"{'Sample Std (ddof=1)':<25} {arr.std(ddof=1):.4f}")
print(f"{'Population Std (ddof=0)':<25} {arr.std():.4f}")
print('='*40)
````

## Problem 5: filtering


### Part 1


Create a 4X6 array with random integers between 0 and 10.  


````python
arr = np.random.randint(0, 11, size = (4, 6))
print(arr)
print(arr.shape)
````

### Part 2


Create a boolean array with `True` entries where the entries in your array are greater than 6


````python
arr = np.arange(30)
arr1 = arr > 6
print(arr[arr1])

print(arr.shape)
````

### Part 3


Use your boolean array as a filter to pull out the entries greater than 6

````python
print(arr[arr > 6])
````
