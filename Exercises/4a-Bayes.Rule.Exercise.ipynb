## Problem 1:

Joan says she is itchy. There is a test for Allergy to Cats, but this test is not always
right:

- For people that really do have the allergy, the test says "Yes" 80% of the time
- For people that do not have the allergy, the test says "Yes" 10% of the time

If 1% of the population has the allergy, and Joan’s test says "Yes", what are the chances that Joan really has the allergy?


```python
# She has a 0.0747 chance or 7.5 % chance of having allergies to cats.

print('='*50)

print(f"Allergies | No Allergies  | Total   | Answer")

allergies =  .80 * 0.01         # This are my positve values

Noallergies =   0.10 * 0.99      # Fasle Positves, 0.10% * 0.99 total poulation
total = allergies + Noallergies
answer = allergies / total
print("="* 50)
print(f'{allergies:^7}   | {Noallergies:7.4f}       |  {total:.4f} | {answer*100:.3f}')
print('='*50)


#print(round(Peoplewithallergies, 4))
#print(Noallergies)
```

## Problem 2:

Beth is planning a picnic today, but the morning is cloudy.

- 50% of all rainy days start off cloudy
- Cloudy mornings are common (40% of all days start cloudy)
- This is usually a dry month (10% of days tend to be rainy)

What is the chance of rain during the day when it starts of cloudy?

```python
#0.1219 OR 12.2%  chance of rain

cloudy_rainy = 0.50   # rain starts of with cloudy days
dry_rainy = .10       # some dry days rain
top = cloudy_rainy * dry_rainy   # mulitplied to .05
cloudy_mornings = .40  # these are days that have clouds but no rain
no_rain = .90         # this days with no rain minus dry rainy days so 100 - 10 leaves 90
x = cloudy_mornings * no_rain  # this value will be added to the total of the top value
total = top + x    # adding top and bottome
answer = top / total  # divided 0.05 by the sum of 0.4100

print('='*40)
print(f"{'Top':<15} | {top:.4f}")
print(f"{'Bottom':<15} | {total:.4f}")
print(f"{'Answer':<15} | {answer*100:.4f}")
print('='*40)
```

## Problem 3:

An artist competition has entries from three painters: Pam, Paul and Peter.

- Pam put in 15 paintings, 4% of her works have won First Prize.
- Paul put in 5 paintings, 5% of his works have won First Prize.
- Peter put in 10 paintings, 3% of his works have won First Prize.

What is the probability that Pam will win First Prize?

```python
#Pam has a 52% chance of winning first prize between the three of them.

all_paintings = 30
pam = 15 / 30         #0.50
paul = 5 / 30         # 0.16666
peter = 10 / 30      # 0.3333333333333333

pam_wins = pam * 0.04   # = 0.0200. This is her winning value  0.04
paul_wins = paul *  0.05  # = 0.0083.  This is his winning value 0.05
peter_wins = peter * 0.03  # = 0.0099.  This is peter's winning value 0.0300

sum_bottom = pam_wins + paul_wins + peter_wins  # add top and bottom values together

answer = pam_wins / sum_bottom  # divide the output of pam_wins by the sum of all bottom values

print('='*45)
print(f"{'Painter':<10} | {'Proportion':^10} | {'Wins':^10}")
print('='*45)
print(f"{'Pam':<10} | {pam:^10.4f} | {pam_wins:^10.4f}")
print(f"{'Paul':<10} | {paul:^10.4f} | {paul_wins:^10.4f}")
print(f"{'Peter':<10} | {peter:^10.4f} | {peter_wins:^10.4f}")
print('='*45)
print(f"{'Total':<10} | {'':^10} | {sum_bottom:^10.4f}")
print('='*45)
print(f"{'P(Pam Wins)':<10} | {answer:^10.4f} | {answer*100:^8.1f}%")
print('='*45)

```
