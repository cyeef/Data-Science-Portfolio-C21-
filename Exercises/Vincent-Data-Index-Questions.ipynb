```python
# --- reset counters before the loop (run this every time) ---
accessible = 0
failed = []

# --- new version with heartbeat ---
for i, (_, row) in enumerate(index.iterrows()):
    if i % 250 == 0:                          # heartbeat: every 250th dataset
        print(f"checked {i} so far...")
    try:
        df = pd.read_csv(row["CSV"])          # try to open it
        accessible += 1                        # success → tally one accessible
    except Exception as e:                     # catch any read failure
        failed.append((row["Item"], type(e).__name__))  # store (name, error type)

print("Total datasets:", len(index))
print("Accessible (readable by Pandas):", accessible)
print("Failed:", len(failed))
for name, err in failed:
    print("   ✗", name, "→", err)
```

Use for loop to enumerate shape, data type, and null counts

```python
records = []
for i, (_, row) in enumerate(index.iterrows()):
    if i % 250 == 0:
        print(f"checked {i}...")
    try:
        df = pd.read_csv(row["CSV"])
        dt = df.dtypes.astype(str).value_counts()      # cast to str so .get works
        records.append({
            "dataset": f"{row['Package']}/{row['Item']}",
            "rows": df.shape[0],
            "columns": df.shape[1],
            "int_cols": int(dt.get("int64", 0)),
            "float_cols": int(dt.get("float64", 0)),
            "object_cols": int(dt.get("object", 0)),
            "cols_with_nulls": int((df.isnull().sum() > 0).sum()),
            "rows_with_nulls": int(df.isnull().any(axis=1).sum()),
            "readable": True,
        })
    except Exception as e:
        records.append({"dataset": f"{row['Package']}/{row['Item']}",
                        "readable": False, "error": type(e).__name__})

profile = pd.DataFrame(records)
profile.to_csv("rdatasets_profile.csv", index=False)   # save it once; it took minutes to build
profile.head(20)
```
