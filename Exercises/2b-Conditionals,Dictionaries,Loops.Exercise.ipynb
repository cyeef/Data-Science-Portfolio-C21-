3. A number raised to the second power is called a square (e.g., $ 3^2 $ is the square of 3). Use a for loop to print out the square of each integer from 1 to 15.

````python
my_num = []  # variable my number with empty list
for n in range(1, 16):  # iterates range betweem 1 and 15 starting at one with the argument 1, and ends at 15
    my_num.append(n ** 2)  # appends number to list after calculation uses method .append()
print(f'The sqaure of each number from 1 to 15 in list {my_num}') # prints formatted string in {my_num}

print("-" * 25)    # prints top line
print(f"{'Number':^10} | {'Squared':^10}")  # prints heading divided by pipe (|)
print("-" * 25)

for n in range(1, 16):  # places each value in the table until the loop terminates
    print(f"{n:^10} | {n ** 2:^10}")  # formats each number and squared value under header Number and Sqauared

print("-" * 25)  # Prints 25 dashes accross
````

4. Write a python program that will take a word and print a message stating whether that word starts with a vowel or a consonant. Test it out with several words.

Example input:

```
popcorn
```

Example output:
```
This word starts with a consonant.
```


````python
vowles = ['a', 'e', 'i', 'o', 'u' ]  # list of vowels
words = ['Green', 'enter','Yellow', 'Brian', 'Susi', 'other', 'alpah', 'instant'] # words
vowles, words  #print variables
````

````python
print("=" * 45)  # print upper border of table using equal sign
print(f"{'This Word':^15} | {'Starts With':^10} | {'Type':^8}")  # this word, starts with and type in heading of table
print("*" * 45) # bottom border of table header stars


for word in words:  # iterates through words by word = one word at a time
  if word[0].lower() in vowles: # looks at first letter of each word and turns it into a lower case
    kind = 'Vowels'  # variable named kind
  else:  # when conditon is false will print
    kind = 'Consonants' # variable assigned to consonants
  print(f"{word:^15} | {word[0].upper():^10} | {kind:^8}")    # formats word, creates uppercase letter and list type vowel or consonant
print("+" * 45)  # bottom boder uses plus sign
````

5. Write a python program that will loop through the following list:
`pets = ['Muffin', 'Kiki', 'Moe', 'Daisy']`. If the length of the pet's name is less than 5 characters, print out the name.  

````python
pets = ['Muffin', 'Kiki', 'Moe', 'Daisy']  #variable and list
pets  # prints names
````

````python
print("$" * 30)   # print top border
print(f"{"Pet Names":<15} | {"Length":<15}") # prints headers used '<' for left margin
print("-"*30)  # bottom line of header

for pet in pets: # iterates through variable pets, checks one name at a time with 'pet'
  if len(pet) < 5: # len() function to determine if word length than five characters
    print(f"{pet:<15} | {len(pet):<15}")  #formats pet individually and gives length

print("="*30)  # prints bottom border
````
