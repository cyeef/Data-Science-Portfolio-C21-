## Use SQL queries to pull specific information

Do NOT pull all the data and then filter using DataFrame methods.  The initial purpose is to get familiar with SQL.




```python
import sqlite3 as db
import pandas as pd
from contextlib import closing
```

```python
 # 11 tables # the internal catalogue of SQLITE - help
# excluding sqllite indexing, triggers and views # sorting by alphabetical order without modificaiton
query = """
  SELECT name
  FROM sqlite_master
  WHERE type = 'table' AND name NOT LIKE 'sqlite_%'
  ORDER BY name;
"""
with closing(db.connect("chinook.db")) as db_con:
    tables = pd.read_sql_query(query, db_con)

tables
```

### Basic Queries


#### SELECT (with * and with column names)


```python
query = """

   SELECT *

  FROM media_types

  LIMIT 10;
"""
with closing(db.connect("chinook.db")) as db_con:
  df = pd.read_sql_query(query, db_con)
df
```

```python
query = """

  Select MediaTypeId, Name

  FROM media_types

  LIMIT 10;

"""

with closing(db.connect("chinook.db")) as db_con:
  df = pd.read_sql_query(query, db_con)

df
```

#### WHERE


```python
query = """

   SELECT *

  FROM invoices

  LIMIT 10;
"""
with closing(db.connect("chinook.db")) as db_con:
  df = pd.read_sql_query(query, db_con)
df
```

```python
#

query = """

  SELECT InvoiceDate

  FROM invoices

  WHERE  Invoicedate LIKE '2009%'

  LIMIT 10;

"""
with closing(db.connect("chinook.db")) as db_con:
    df = pd.read_sql_query(query, db_con)
df
```

```python
query = """

  SELECT InvoiceDate

  FROM invoices

  WHERE  strftime('%Y', Invoicedate) = "2009"

  LIMIT 15;

"""
with closing(db.connect("chinook.db")) as db_con:
    df = pd.read_sql_query(query, db_con)
df
```

#### AND


```python
query = """

  SELECT BillingCity, BillingPostalCode, BillingCity, TOTAL

  FROM invoices

  WHERE  BillingCity = 'Stuttgart'
   AND BillingPostalCode = '70174'
   AND  BillingState IS NULL

  LIMIT 15;

"""
with closing(db.connect("chinook.db")) as db_con:
    df = pd.read_sql_query(query, db_con)
df
```

#### OR


```python
query = """

  SELECT BillingCity, BillingPostalCode, BillingCity, TOTAL

  FROM invoices

  WHERE  BillingCity = 'Stuttgart' OR 'Frankfurt'
   AND BillingPostalCode = '70174'
    OR BillingState IS NULL

  LIMIT 15;

"""
with closing(db.connect("chinook.db")) as db_con:
    df = pd.read_sql_query(query, db_con)
df
```
