## Part 1: Numbers (Integers & Floats)

```python
# 1. Doubling integers
numbers = [1, 2, 3, 4, 5]
doubled = [x * 2 for x in numbers]
doubled
```

```python
# 1. Doubling integers

numbers = [1, 2, 3, 4, 5]
doubled = []

for hy in numbers:
  doubled.append(hy * 2)
print(doubled)
```

```python
# subtracting integers by 1

graze = [ 23, 24, 26, 28 , 30]
minime = [ x - 1 for x in graze]
print(minime)
```

```python
minime = []

for gee in graze:
  minime.append(gee - 1)
print(minime)
```

```python
# adding intergers by 1
gogo = [ 2, 7, 9, 34, 9 ,0]
addy = [dig + 1 for dig in gogo]
print(addy)
```

```python
addy = []

for dig in gogo:
  addy.append(dig + 1)

print(addy)
```

```python
# 2. Squaring numbers
data = [1, 3, 5, 7, 9]
squares = [num ** 2 for num in data]
```

```python
# Your for-loop here:
data = [1, 3, 5, 7, 9]
squares = []
for bingo in data:
  squares.append(bingo ** 2)
print(squares)
```

```python
# 3. Converting Celsius to Fahrenheit
celsius = [0.0, 10.5, 20.0, 35.6]
fahrenheit = [(temp * 9/5) + 32 for temp in celsius]
print(fahrenheit)
```

```python
# Your for-loop here:
fahrenheit = []
for tazz in celsius:
  fahrenheit.append(tazz * 9/5 + 32)
print(fahrenheit)
```

```python
# Converting Fahrenheit to Celsius
fahrenheit = [ 31, 66, 89.5, 99, 32]
celsius = [round((yuppi - 32) * 5/9, 3) for yuppi in fahrenheit]
print(celsius)
```

```python
# 4. Filtering for even numbers
scores = [12, 15, 18, 21, 24, 27, 30]
evens = [n for n in scores if n % 2 == 0]
print(evens)
```

```python
# Your for-loop here:
evens = []
for eck in scores:
  if eck % 2 == 0:
    evens.append(eck)
print(evens)
```

```python
# Filtering for odd numbers
scores = [ 26, 345, 67, 4, 9, 5]
odds = [yammy for yammy in scores if yammy % 2 != 0]
print(sorted(odds, reverse = False))
```

```python

odds = []
for juni in scores:
  if juni % 2 != 0:
    odds.append(juni)
print(sorted(odds, reverse = False))
```

## Part 2: Strings

```python
# 6. Uppercasing names
names = ["alice", "bob", "charlie"]
shouting = [name.upper() for name in names]
print(shouting)
```

```python
# Your for-loop here:
shouting = []
for ju in names:
  shouting.append(ju.upper())
print(shouting)
```

```python
shouting = [ ju.title() for ju in names]
print(shouting)
```

```python
# 7. Getting string lengths
words = ["apple", "banana", "cherry", "date"]
lengths = [len(w) for w in words]
print(lengths)
```

```python
# Your for-loop here:
lengths = []
for yack in words:
  lengths.append(len(yack))
print(lengths)
```

```python
# 8. Filtering by initial letter
animals = ["cat", "dog", "cheetah", "elephant", "camel"]
c_animals = [animal for animal in animals if animal.startswith("c")]
print(c_animals)
```

```python
# Your for-loop here:
c_animals = []

for ghi in animals:
  if ghi.startswith('c'):
    c_animals.append(ghi)
print(c_animals)
```

## Part 3: Mixed & Combined Concepts

```python
# 9. Finding long words
fruits = ["kiwi", "mango", "watermelon", "fig", "pineapple"]
long_fruits = [f for f in fruits if len(f) > 5]
```

```python
# Your for-loop here:
for had in fruits:
  if len(had) > 5:
    long_fruits.append(had)
print(long_fruits)
```

```python
# 10. Halving numbers if they are large
weights = [10.0, 50.5, 4.2, 100.0, 12.8]
halved_large = [w / 2 for w in weights if w > 15.0]
```

```python
# Your for-loop here:
for uy in weights:
  if uy > 15.0:
     halved_large.append(uy)
print(halved_large)
```
