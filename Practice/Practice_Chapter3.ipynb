## Section 3.4 Control Statements

*Sequential execution* is the order in which a program executes.  You can write the program so the order of execution is changes- *transfer of control*.  

### Selection Statements
Conditions can determine the flow of code by evaluated expressions to be **True** or **False**

-*if* statement evaluates a True condition it executes that statement or terminates if the condition is false

- *if else* can perform a condition if true or do something else if false

- *if elif else *statement can execute multiple statements based upon the evaluated conditions assigned - if True do this, if false do this or do this if the evaluation is neither true or false



## Repetition Statements
